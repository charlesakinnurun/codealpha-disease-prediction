# 01 — Heart Disease EDA

Exploratory analysis of the UCI Cleveland heart disease dataset. This notebook
is a thin wrapper around the tested package functions so the analysis stays
reproducible from the command line (`python scripts/validate_data.py`).

**Educational/research use only — not clinical evidence.**

```python
from disease_prediction.config import get_settings
from disease_prediction.data.loaders import load_dataset, target_counts
from disease_prediction.data.registry import get_schema
from disease_prediction.data.validate import validate_dataframe

task = "heart_disease"
settings = get_settings()
schema = get_schema(task)
df = load_dataset(task)
df.shape, target_counts(df)
```

```python
report = validate_dataframe(df, schema)
print(report.format_text())
```

```python
df[schema.numeric_features()].describe().T
```

```python
# Missingness: ca and thal carry '?' in the raw file, preserved as NaN.
df[schema.feature_names].isna().sum().sort_values(ascending=False).head(5)
```

```python
df.groupby("target")[["age", "trestbps", "chol", "thalach", "oldpeak"]].mean()
```

```python
# Regenerate the standard figure set into reports/figures/heart_disease/
from disease_prediction.data.eda import generate_eda_figures

for path in generate_eda_figures(task, df, schema, settings.figures_dir):
    print(path)
```

```python
# Correlation structure (numeric features only).
df[schema.numeric_features()].corr().round(2)
```
