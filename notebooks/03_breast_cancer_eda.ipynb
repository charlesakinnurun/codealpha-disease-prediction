# 03 — Breast Cancer (WDBC) EDA

Exploratory analysis of the Breast Cancer Wisconsin (Diagnostic) dataset,
loaded from scikit-learn's bundled copy of the UCI data. The positive class is
**malignant**.

**Educational/research use only — not clinical evidence.**

```python
from disease_prediction.config import get_settings
from disease_prediction.data.loaders import load_dataset, target_counts
from disease_prediction.data.registry import get_schema
from disease_prediction.data.validate import validate_dataframe

task = "breast_cancer"
settings = get_settings()
schema = get_schema(task)
df = load_dataset(task)
df.shape, target_counts(df)
```

```python
print(validate_dataframe(df, schema).format_text())
```

```python
# The mean/error/worst triplets are strongly correlated, which makes
# single-feature attributions unstable — see docs/model_card.md.
import pandas as pd

corr = df[schema.feature_names].corr().abs()
pairs = corr.where(~(corr >= 1.0)).stack().sort_values(ascending=False).head(10)
pd.DataFrame({"pair": pairs.index.tolist(), "abs_corr": pairs.values})
```

```python
df.groupby("target")[["mean radius", "mean concavity", "worst area", "mean texture"]].mean()
```
