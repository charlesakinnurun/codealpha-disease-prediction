# 02 — Diabetes EDA

Exploratory analysis of the UCI Early Stage Diabetes Risk Prediction dataset.
The label is a **questionnaire-based risk classification**, not a laboratory
diagnosis. Run `python scripts/download_data.py --task diabetes` first.

**Educational/research use only.**

```python
from disease_prediction.config import get_settings
from disease_prediction.data.loaders import diabetes_raw_path, load_dataset, target_counts
from disease_prediction.data.registry import get_schema
from disease_prediction.data.validate import validate_dataframe

task = "diabetes"
settings = get_settings()
schema = get_schema(task)
assert diabetes_raw_path().exists(), "run scripts/download_data.py --task diabetes"
df = load_dataset(task)
df.shape, target_counts(df)
```

```python
print(validate_dataframe(df, schema).format_text())
# Note the large number of duplicate feature rows — a documented property of
# this questionnaire dataset that limits how much signal a model can extract.
```

```python
df["Age"].describe(), df["Gender"].value_counts()
```

```python
# Symptom prevalence by class (binary features are 0/1).
binary = schema.binary_features()
df.groupby("target")[binary].mean().T.sort_values(1, ascending=False).head(10)
```

```python
# Age distribution by class and the standard figure set.
import matplotlib.pyplot as plt

from disease_prediction.data.eda import generate_eda_figures

ax = df.boxplot(column="Age", by="target")
ax.set_title("Age by target")
plt.suptitle("")
plt.show()

for path in generate_eda_figures(task, df, schema, settings.figures_dir):
    print(path)
```
